# Signal — Fintech & Credit-Risk News Digest Agent

## What it does, and for whom

Signal is a small research agent that scans for genuinely notable fintech and
credit-risk news — fraud detection, underwriting, embedded finance, lending
regulation, African fintech specifically — and turns it into a structured,
4-story digest. It's built for one person: me, staying current on the
industry I'm job-hunting in, without spending an hour a week manually
scanning ten sites. It doubles as a live demonstration piece for my
portfolio, showing a working AI tool rather than just describing one.

## Setup (a stranger could follow this)

1. You need access to a Claude.ai account with Artifacts enabled — this tool
   runs as a self-contained HTML artifact, not a standalone hosted app (see
   Limitations for why).
2. Open `signal_digest_agent.html` as an artifact in a Claude conversation
   (paste the code into a new artifact, or ask Claude to render it).
3. No API key setup needed — inside the Claude artifact sandbox, API calls
   are proxied automatically.
4. No other accounts, installs, or configuration required.

## Usage example

1. Click **"Generate this week's digest."**
2. Wait a few seconds while it searches live.
3. A 4-story digest appears, each with a headline, a one-sentence "why it
   matters," and a source link.
4. Every digest generated is saved automatically — the left-hand archive
   fills in over time, and you can click any past date to reopen it.

## Architecture (simple sketch)

```
[Your click]
     │
     ▼
[Browser JS: builds the prompt]
     │
     ▼
[Claude API, with web_search tool enabled]
     │  (Anthropic's server runs the actual search,
     │   returns results + Claude's synthesis in one call)
     ▼
[JSON digest returned to the browser]
     │
     ▼
[Parsed + rendered as story cards]
     │
     ▼
[Saved to window.storage, keyed by today's date]
     │
     ▼
[Archive list re-rendered so it's browsable later]
```

No custom backend exists. The "backend" work — running the search and
writing the digest — happens inside Anthropic's API call itself; the browser
only handles display and local saving.

## v2 Evaluation Results

See `EVAL_RESULTS.md` — summary: 2 of 2 real trial runs produced valid,
specific, non-duplicate 4-story digests. The two real weaknesses found were
stale stories passed off as "this week's" (recency isn't actually enforced)
and one source link resolving to a category page instead of a specific
article. Full detail, including which stories were selected and why, is in
that file.

## Limitations

See `LIMITATIONS.md` for the full list. The two that matter most if you're
deciding whether to trust a digest as-is: recency isn't verified by the tool
itself, and it can't run on a schedule — it's triggered by hand, and the
"recurring" part only comes from the saved archive, not automation.
