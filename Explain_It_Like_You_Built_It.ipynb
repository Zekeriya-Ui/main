### Assignment: Explain It Like You Built It

**Track:** General AI Fluency (Week 6)

**My Build Piece:** Automated Continuous Deployment (Git → GitHub → Host)

**Explanation:**

Before setting up deployment, I thought making a website live meant manually uploading HTML/CSS files through a browser every single time I tweaked a pixel or fixed a typo. Instead, my site uses an automated pipeline between GitHub and my hosting platform.

Imagine GitHub as a secure digital vault where I store the master blueprints of my code. My host acts like an automated construction worker standing right next to that vault, constantly watching for updates.

Whenever I save my work locally and run `git push`, I'm sending the newest blueprint to the vault. The instant the host detects a fresh commit, its automated background systems wake up, pull a copy of the code, verify everything builds correctly, and swap out the old live site for the new version behind the scenes. The entire pipeline finishes in under 30 seconds without me ever touching a manual server console.
