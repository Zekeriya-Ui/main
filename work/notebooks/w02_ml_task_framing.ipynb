## 1. My lane as an ML task (type)

**Task Type:** Multi-class Classification

**Why:** We are predicting which of three distinct species a flower belongs to based on physical measurements. Since there are more than two categories and no inherent order, this is a multi-class classification problem.

```python
from sklearn.datasets import load_iris
import pandas as pd

# Load dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Check distribution
target_counts = df['species'].value_counts(normalize=True)
print("Target Distribution (Species 0, 1, 2):")
print(target_counts)
```

## 5. Why ML beats a fixed rule here

While species 0 is easily separated by a simple rule (e.g., petal length < 2), species 1 and 2 overlap significantly in their measurements. A fixed if-statement would struggle to find the optimal boundary that minimizes error across all four dimensions, whereas a model like a Decision Tree or Logistic Regression can optimize this boundary mathematically.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(df.iloc[:, 2], df.iloc[:, 3], c=df['species'], cmap='viridis')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Feature Overlap between Species')
plt.show()
```
