## 3. Human review + the no-go list

Before any action is taken based on the playbook's recommendations, human review is crucial to ensure alignment with brand guidelines, current campaigns, and unforeseen external factors. Reviewers must exercise qualitative judgment, especially for high-impact or sensitive content.

**What a person must check before acting:**
*   **Brand Alignment:** Does the recommended action align with current brand messaging and tone?
*   **Campaign Conflicts:** Does the action conflict with any ongoing marketing campaigns or editorial calendars?
*   **Sensitivity:** Is the content or recommended action sensitive, requiring a more nuanced approach than the model suggests?
*   **Legal/Compliance:** Are there any legal or compliance implications for the suggested action (e.g., copyright, data privacy)?
*   **Contextual Relevance:** Is the recommendation still relevant given any recent, unforeseen events?

**What should never be automated (the no-go list):**
*   **Deletion of Core Content:** Never automatically delete foundational or evergreen content without explicit human approval.
*   **Public-Facing Tone Changes:** Major changes to public-facing content tone or voice should always involve human review.
*   **Content with Legal Implications:** Any content touching on legal, financial, or health advice must undergo strict human scrutiny.
*   **Crisis Communications:** Automated actions during a crisis can be disastrous; all communications must be human-approved.
*   **Unusual Spikes/Dips:** Recommendations based on anomalous data spikes or dips should be reviewed to understand the root cause before action.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Example: Flagging actions that require higher human review based on keywords or predicted impact

def flag_for_review(action_row):
    if action_row['predicted_impact'] > 0.90 or 'sensitive' in action_row['reason_code'].lower():
        return 'High Priority Review'
    elif 'deletion' in action_row['action_type'].lower() or 'archive' in action_row['action_type'].lower():
        return 'Mandatory Review (No-Go List Item)'
    return 'Standard Review'

# Assuming 'ranked_actions_df' from section 1 is available
if 'ranked_actions_df' in locals():
    ranked_actions_df['review_level'] = ranked_actions_df.apply(flag_for_review, axis=1)
    display(ranked_actions_df)
else:
    print("Run Section 1's code to get 'ranked_actions_df' first.")

# This cell could also contain queries to a content management system to identify
```

## 4. Monitoring / retrain triggers

To ensure the action playbook remains effective and its recommendations accurate, continuous monitoring of key performance indicators (KPIs) and model performance metrics is essential. Specific triggers will indicate when the model needs retraining or the playbook requires an update.

**What would tell you the recommendations went stale?**
*   **Decreased Action Effectiveness:** A significant drop in the actual impact of recommended actions (e.g., lower-than-predicted engagement for promoted articles).
*   **Prediction Drift:** The model's predictions consistently diverge from actual outcomes, indicating data drift or concept drift.
*   **User Feedback:** Increased negative feedback from content strategists or editors regarding the relevance or utility of recommendations.
*   **Key Metric Decline:** A sustained decline in overall content-related KPIs (e.g., total engagement, conversion rates, time on page).
*   **Data Schema Changes:** Any changes in the upstream data sources or content platform schema that the model relies upon.
*   **External Market Shifts:** Major industry events, competitor actions, or changes in audience behavior not captured by the current model.

**Retrain Triggers:**
*   If any of the above monitoring indicators exceed predefined thresholds (e.g., 10% drop in prediction accuracy).
*   Scheduled retraining (e.g., quarterly or semi-annually) to incorporate new trends and data.
*   Significant updates to the content catalog or introduction of new content types.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Placeholder for monitoring metrics
import numpy as np

# Simulate some monitoring metrics
actual_impact = np.array([0.90, 0.85, 0.70]) # Actual impact of actions
predicted_impact_baseline = np.array([0.95, 0.88, 0.75]) # Model's predicted impact from Section 1

# Calculate a simple effectiveness metric (e.g., difference)
effectiveness_delta = predicted_impact_baseline - actual_impact

# Define a threshold for staleness
staleness_threshold = 0.10 # 10% deviation

if np.any(effectiveness_delta > staleness_threshold):
    print("WARNING: Some actions show effectiveness significantly lower than predicted.")
    print(f"Differences: {effectiveness_delta}")
    print("Retraining or deeper investigation may be required.")
else:
    print("Action effectiveness within expected bounds. No immediate retraining trigger.")

# This cell would ideally connect to a monitoring dashboard or data warehouse to pull actual performance data.
```
