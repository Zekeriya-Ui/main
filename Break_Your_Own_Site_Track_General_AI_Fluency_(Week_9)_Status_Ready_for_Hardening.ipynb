2. Evidence of Fixes Addressed ("Fix-Nows")
*   **Input Sanitization & Validation:**
    *   **Fix:** Implemented front-end schema validation (`trim()`, maximum length checks, and escaping HTML entities before rendering user input).
*   **Double-Submit Prevention:**
    *   **Fix:** Added a `isSubmitting` state toggle that disables the submit button immediately upon invocation and displays a loading spinner.
*   **Mobile & Viewport Layout Overflows:**
    *   **Fix:** Adjusted CSS media queries for `<480px` viewports, wrapped flex elements with `flex-wrap: wrap`, and added `overflow-x: hidden` on main containers.
*   **404 Link Repairs:**
    *   **Fix:** Audited all `href` attributes across external links and verified proper repository URLs and `rel="noopener noreferrer"` security flags.
