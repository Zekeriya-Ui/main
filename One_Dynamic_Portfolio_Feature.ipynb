### Step 2: Implementation & Wiring Guide (Option A: Serverless Contact Form)

#### 1. The Front-End Markup (HTML)

Add this clean, responsive form code to the #contact section of your portfolio HTML:

```html
<form id="portfolio-contact" action="https://formspree.io/f/YOUR_FORM_ID" method="POST" class="contact-form">
  <div class="form-group">
    <label for="name">Name</label>
    <input type="text" id="name" name="name" required placeholder="Jane Doe" />
  </div>
  
  <div class="form-group">
    <label for="email">Email</label>
    <input type="email" id="email" name="email" required placeholder="jane@example.com" />
  </div>
  
  <div class="form-group">
    <label for="message">Message / Technical Inquiry</label>
    <textarea id="message" name="message" rows="4" required placeholder="Let's discuss risk models or technical assessments..."></textarea>
  </div>

  <button type="submit" class="cta-button">Send Message</button>
</form>
```

#### 2. The Styling (CSS)

Add this block to your style tag to match your dark-theme portfolio palette:

```css
.contact-form {
  display: grid;
  gap: 16px;
  margin-top: 18px;
  max-width: 500px;
}
.form-group {
  display: flex;
  flex-direction: column;
  gap: 6px;
}
.form-group label {
  font-family: 'IBM Plex Mono', monospace;
  font-size: 12px;
  color: var(--text-muted);
}
.form-group input,
.form-group textarea {
  background: var(--surface-raised);
  border: 1px solid var(--hairline);
  border-radius: 8px;
  padding: 12px;
  color: var(--text);
  font-family: 'Inter', sans-serif;
  font-size: 14px;
}
.form-group input:focus,
.form-group textarea:focus {
  outline: 1px solid var(--accent);
  border-color: var(--accent);
}
```

**Assignment:** One Dynamic Feature

#### 1. What a Backend Is:

A front-end web page is like the menu and dining room of a restaurant—it's everything the visitor sees and clicks on. A backend is the kitchen in the back. It’s the invisible logic, server hardware, and database systems that take requests from the front-end, process information securely, and trigger actions (like storing data, running calculations, or sending notifications).

#### 2. What My Feature Does:

Instead of forcing visitors to copy my email address or opening an external mail app, my site has an inline form. A recruiter or risk engineering team can type a message directly on the page, hit "Send Message," and receive instant visual confirmation on screen.

#### 3. How the Data Flows (End-to-End):

**User Input:** The visitor types their name, email, and message into HTML input fields and clicks "Submit."

**HTTP POST Request:** The browser packages those inputs into a digital envelope (a POST request) and sends it over the internet to my serverless endpoint URL.

**Backend Processing:** Formspree's cloud server receives the payload, verifies that it isn't automated spam, and formats the raw form data into an email body.

**Delivery:** The backend server delivers the formatted message straight to my personal email inbox within seconds, closing the loop.
