## Content Map

### 1. Landing Page
| Order | Section | Content |
|---|---|---|
| 1 | Hero | The one-line claim, full stop — no supporting paragraph crowding it |
| 2 | Proof teaser | One real number from the case study (e.g. "2.5x lift in the riskiest decile"), linking straight to Case Studies |
| 3 | One-line bio | "Five years in credit operations, MSc Financial Engineering, ML training at FlyRank AI." Nothing more — full bio lives on Methodology |
| **CTA** | **"See the proof"** → links to Case Studies (soft CTA — landing's job is to earn the click, not close the ask) |

### 2. Case Studies / Work
| Order | Section | Content |
|---|---|---|
| 1 | Case 1 (lead, only finished case) — The problem | CV can't prove I can code a working model; this exists so a hiring manager can check before reading a resume line |
| 2 | What I did & decided | Observation/outcome window split; the AUC=1.0 leakage catch and fix — the actual evidence of rigor |
| 3 | What came of it | AUC 0.679, KS 0.319, decile lift chart; stated plainly as synthetic-data results, not real portfolio numbers |
| 4 | Reproducibility | Link to notebook/repo (**not yet live — see gather list**) |
| **CTA** | **"Invite me to your assessment"** → direct ask, placed here because this is where trust is actually built |

### 3. About / Methodology
| Order | Section | Content |
|---|---|---|
| 1 | Approach | How I handle imbalanced data, leakage checks, validation choices (AUC/KS/lift over accuracy) — judgment, not biography |
| 2 | Credentials | One line: Watu Credit, MSc Financial Engineering, FlyRank ML training — no expanded bio |
| **CTA** | **"Invite me to your assessment"** → same ask, reinforcing, not a new one |

### 4. Contact
| Order | Section | Content |
|---|---|---|
| 1 | Restated ask | "I'd rather you test this than take my word for it." |
| 2 | Contact method | Email / LinkedIn — **decision still needed, see gather list** |
| **CTA** | **"Invite me to your assessment"** → the literal execution of the one action |

**CTA ladder check:** Landing's soft CTA ("see the proof") leads to Work, where the real ask first appears and repeats identically through About and Contact. No page introduces a competing or softer ask later in the funnel — the assessment invite is the only action anywhere past the landing page.
