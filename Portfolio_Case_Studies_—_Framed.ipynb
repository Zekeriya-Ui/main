## The Framed Case Study

### The problem
My CV lists credit operations experience and ML credentials side by side. It doesn't prove I can build a model that works. I needed something a Head of Risk Analytics could open and check before they ever read a resume line.

### What I did — and what I decided
I built an early-stage delinquency model: using behavior from a loan's first three installments to predict delinquency in the next three. Partway through, the first version scored a perfect AUC of 1.0 — not a win, a leak. The features and the label were built from the same window, so the model was just repeating the answer back to me. I rebuilt the target definition to split observation from outcome, and re-validated from scratch.

### What came of it
AUC 0.679, KS 0.319, and a decile lift chart showing the riskiest 10% of clients flagged at 2.5x the base rate — real numbers, not the fake 1.0. It's built on a synthetic dataset matched to Home Credit's real schema, and I say so directly, because the model isn't the point. Catching my own mistake before it shipped is.
