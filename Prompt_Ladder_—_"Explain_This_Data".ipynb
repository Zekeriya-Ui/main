# Prompt Ladder — "Explain This Data"

## Source task:
Explaining my case-study model's validation output — the kind of prompt I'd have typed a month ago, pointing at a results table with no further thought.

## The data being explained, in every version below (kept identical so only the prompt changes):
```
AUC-ROC: 0.679
KS statistic: 0.319
Average precision: 0.181
Base rate (test set): 0.090

Decile lift table (riskiest decile first):
Decile 1: 2.50x lift
Decile 2: 1.67x lift
Decile 3: 1.61x lift
...
Decile 10: 0.44x lift

Top permutation-importance features:
EXT_SOURCE_2: 0.0677
EXT_SOURCE_1: 0.0393
utilization_trend: 0.0083
delay_trend: 0.0076
```

---

## V0 — Baseline (the embarrassing one)

**Prompt:**
> Explain this data.
> [metrics above]

**Output (excerpt):**
> This data shows a machine learning model's performance metrics. The AUC-ROC of 0.679 indicates moderate discriminative ability between classes. The KS statistic of 0.319 suggests reasonable separation between positive and negative distributions. Average precision of 0.181 reflects performance under class imbalance. The decile lift table shows the model concentrates positive outcomes in higher-risk deciles, with the top decile showing 2.5x lift over the base rate. Feature importance indicates EXT_SOURCE_2 and EXT_SOURCE_1 are the strongest predictors, followed by behavioral features like utilization_trend and delay_trend.

**Notes:**
- *What changed in the prompt:* Nothing — two words, no goal, audience, or context.
- *What actually improved in the output:* Nothing to compare against yet — this is the floor.
- *What still failed:* It's a metric glossary, not an explanation of *my* data. It would say almost exactly this for any AUC/KS/lift table from any model, mine or a stranger's. It doesn't say whether 0.679 is good, bad, or a red flag — the actual question I need answered.
- *What I'd try next:* Give it a reason to have an opinion — a goal.

---

## V1 — Add: A Clearer Goal

**Prompt:**
> Explain this data so I can decide whether it's strong enough to include in my job-search portfolio.
> [metrics above]

**Output (excerpt):**
> Given your goal of using this in a portfolio, here's the honest read: an AUC of 0.679 and KS of 0.319 are respectable for an early-warning model, not spectacular — this won't win a Kaggle leaderboard, but portfolio pieces aren't scored like competitions. What matters more is the decile lift: your top-risk decile is catching outcomes at 2.5x the base rate, which is a genuinely useful operational number if this were deployed. I'd say yes, include it, but the story should be "here's a working early-warning signal," not "here's a best-in-class model."

**Notes:**
- *What changed in the prompt:* Added a goal — "so I can decide whether it's strong enough to include."
- *What actually improved in the output:* It stopped just describing the numbers and started rendering a verdict — "respectable, not spectacular" — and reframed what "good" even means for a portfolio piece versus a competition. That reframe didn't exist in V0 at all.
- *What still failed:* It doesn't know who's reading the portfolio. "Respectable for an early-warning model" is a judgment aimed at nobody in particular — a recruiter and a risk quant would weigh that verdict completely differently, and the explanation doesn't adapt to either.
- *What I'd try next:* Name the actual audience.

## V2 — Add: A Defined Audience

**Prompt:**
> Explain this data so I can decide whether it's strong enough to include in my job-search portfolio. The audience reading it is a Head of Risk Analytics at a fintech — someone who evaluates credit models daily and will not be impressed by generic ML vocabulary.
> [metrics above]

**Output (excerpt):**
> For that audience, lead with the lift table, not AUC — a risk analytics lead thinks in terms of "what happens if we act on the top decile," and 2.5x concentration there is the number that actually maps to an operational decision, not 0.679 in isolation. AUC and KS are fine as supporting evidence, but citing them first reads as "I know the vocabulary" rather than "I know how this gets used." One thing to be upfront about to this specific audience: 0.679 alone, without knowing the base rate context (9%), can look unremarkable — you need to state the base rate right next to it or a sharp reader will wonder if you're hiding something.

**Notes:**
- *What changed in the prompt:* Named the specific audience and what would and wouldn't impress them.
- *What actually improved in the output:* It stopped treating AUC as the headline number and reordered the whole explanation around what a risk analyst actually decides with — the lift table. It also proactively flagged a credibility risk (stating AUC without base rate looks evasive) that neither V0 nor V1 mentioned at all, because neither had a skeptical reader in mind.
- *What still failed:* It's reasoning about a generic "Head of Risk Analytics," not about *my specific pipeline* — it doesn't know about the target-leakage bug I caught and fixed, or the observation/outcome window split, which are the actual differentiators of this project. It's giving good generic advice for *a* risk model, not *this* one.
- *What I'd try next:* Give it the real background it's missing.

## V3 — Add: Real Context

**Prompt:**
> Explain this data so I can decide whether it's strong enough to include in my job-search portfolio. The audience is a Head of Risk Analytics at a fintech who evaluates credit models daily. Context: this is an early-stage delinquency model — features from installments 1-3 predict delinquency in installments 4-6 (a deliberate observation/outcome window split). An earlier version leaked the label (features and target shared the same window) and produced AUC 1.0; this is the corrected, non-leaked version.
> [metrics above]

**Output (excerpt):**
> That leakage history is actually your strongest asset here, and it should be in the portfolio text, not left out. A Head of Risk Analytics has almost certainly seen an AUC of 1.0 before and knows exactly what it means — not "great model," but "broken evaluation." Showing that you caught it, understood why it happened (window overlap between feature and target), and corrected it demonstrates the specific skill this hire is actually for: knowing when a result is too good to be true. The corrected 0.679/0.319 numbers, read next to that story, land as evidence of rigor rather than a mediocre score to be talked around.

**Notes:**
- *What changed in the prompt:* Added the real pipeline background — the window split and the leakage bug — instead of leaving the model as an anonymous data table.
- *What actually improved in the output:* This is the first version to treat the leakage catch as the headline, not a footnote — it explicitly reframed a "weaker-looking number" as the actual proof point, which none of the earlier versions could do because they didn't know that story existed.
- *What still failed:* The explanation is now a paragraph of prose. If this is actually going on a portfolio page, I need something I can drop into a fixed layout, not free-flowing text I have to manually restructure every time I regenerate it.
- *What I'd try next:* Constrain the output format.

## V4 — Add: An Example of What Good Looks Like

**Prompt:**
> Explain this data so I can decide whether it's strong enough to include in my job-search portfolio. Audience: Head of Risk Analytics at a fintech. Context: early-stage delinquency model, installments 1-3 predict installments 4-6, an earlier leaked version scored AUC 1.0 and was corrected. Here's an example of a well-written metric explanation from an unrelated field, for style reference:
> "Our A/B test ran for 14 days across 40,000 users. The 3.2% lift in signup conversion is statistically significant (p=0.02), but the practical impact is modest — at current traffic, that's roughly 60 extra signups per week, not the 500 we'd need to justify the engineering cost of a full rollout."
> Write in that style.
> [metrics above]

**Output (excerpt):**
> Our model was validated on a held-out set of 2,000 clients. The AUC of 0.679 is statistically above chance (base rate 9%), but the practical impact is what matters — the top-risk decile catches delinquency at 2.5x the base rate, meaning if a risk team reviewed just the top 10% flagged, they'd catch roughly 2.5x more early-warning cases than reviewing a random 10% of the portfolio. That's meaningful triage value, though not a silver bullet — deciles 4 through 10 show lift near or below 1.0, so the model's edge is concentrated at the top, not evenly spread.

**Notes:**
- *What changed in the prompt:* Added a worked example from an unrelated domain (an A/B test writeup) as a style reference.
- *What actually improved in the output:* Almost nothing beyond V3 — it adopted the example's "statistically X, but practically Y" sentence pattern, which is fine, but it also invented a specific number ("2,000 clients") that wasn't in my data or context, apparently to mimic the example's concrete-sample-size structure. That's a real regression, not a neutral wash.
- *What still failed:* This is the honest "didn't help" moment — the example nudged style at the cost of accuracy. Because the reference example had a specific, concrete sample size stated, the model pattern-matched that slot and filled it with something plausible-sounding rather than something true. A stylistic example is risky exactly when the model treats its specific *content*, not just its *shape*, as something to imitate.
- *What I'd try next:* Drop the stylistic example, and instead add an explicit instruction to verify every number against the source data before stating it.

## V5 — Add: A Verification Requirement

**Prompt:**
> Explain this data so I can decide whether it's strong enough to include in my job-search portfolio. Audience: Head of Risk Analytics at a fintech. Context: early-stage delinquency model, installments 1-3 predict installments 4-6, an earlier leaked version scored AUC 1.0 and was corrected. Format: 3 short paragraphs — (1) the headline number and what it means operationally, (2) the leakage-catch story as evidence of rigor, (3) one honest limitation. Before finalizing, check every specific number you state against the data block below and remove or flag anything not explicitly present in it.
> [metrics above]

**Output:**
> The number that matters operationally is the decile lift, not AUC in isolation: the top-risk decile catches delinquency at 2.5x the portfolio's base rate of 9%, meaning a risk team reviewing just the riskiest 10% would catch a disproportionate share of near-term delinquency before it compounds.
>
> The more interesting evidence of rigor is what got fixed along the way: an earlier version of this model, which computed behavior features and the target from the same installment window, scored a suspicious AUC of 1.0 — a textbook sign of label leakage, not skill. Splitting the observation window (installments 1-3) from the outcome window (installments 4-6) corrected this, and the resulting AUC of 0.679 and KS of 0.319 are the honest, non-leaked numbers.
>
> One limitation worth stating plainly: lift declines unevenly past the top decile (deciles 4-10 hover near or below 1.0x), so the model's value is concentrated at the top of the risk distribution, not spread evenly — a real constraint on how it should be used operationally, not a flaw to hide.

**Notes:**
- *What changed in the prompt:* Added an explicit format (3 fixed paragraphs) and a verification instruction to check every number against the source data before finalizing.
- *What actually improved in the output:* No invented numbers this time — every figure stated (2.5x, 9%, 0.679, 0.319, "deciles 4-10 near or below 1.0x") traces directly back to the data block, unlike V4's fabricated sample size. The output is also now a fixed, three-part structure I can drop straight onto a portfolio page without editing.
- *What still failed:* It's now genuinely close to done for this specific use case — the remaining gap is that it's still one draft; nothing forces a second pass checking whether the tone matches how I'd actually say this out loud.
- *What I'd try next:* If iterating further, I'd add a review-instruction layer ("read this back as if you were the skeptical reader and note anything you wouldn't believe") — but for this deliverable, V5 is the stopping point.

## Final Reusable Prompt

```
Explain [DATA/RESULTS] so I can decide whether [DECISION THIS EXPLANATION
NEEDS TO SUPPORT, e.g. "it's strong enough to include in my portfolio"].

Audience: [SPECIFIC PERSON/ROLE] who [WHAT WOULD OR WOULDN'T IMPRESS THEM
— avoid generic praise-language here].

Context: [THE REAL BACKGROUND A STRANGER WOULDN'T KNOW — how this was
built, any mistake that was caught and fixed, anything that makes this
specific rather than generic].

Format: [FIXED STRUCTURE, e.g. "3 short paragraphs: headline number and
what it means operationally, one piece of evidence of rigor, one honest
limitation"].

Before finalizing, check every specific number or claim you state against
the source data below, and remove or explicitly flag anything not present
in it.

[PASTE SOURCE DATA HERE]
```

**Why this works for a stranger:** every bracket is a variable, not a fact about credit risk or me specifically — the structure (goal → audience → real context → fixed format → verification check) forces the same discipline regardless of what data or decision sits inside it. The verification line is the one non-negotiable part: it's the layer that turned V4's plausible-sounding fabrication into V5's checked, honest output, and that risk exists for any dataset, not just this one.
