## A. 5-Minute Showcase Demo Outline

*   **0:00 - 1:00 | The Problem (FlyRank Case Study):**
    *   *Question:* How can search platforms predict engagement and rerank content across 79M query signals without exposing raw query text or private client identifiers?
    *   *Constraint:* High-cardinality cold-start terms break naive text lookups and present data privacy risks.

*   **1:00 - 2:00 | Methodology & Architecture:**
    *   *Dataset:* 79 million production search interactions across a 21-day temporal window.
    *   *Pipeline:* Temporal train/validation split (Days 1–17 train, Days 18–21 temporal holdout) with zero-leakage feature engineering (session depth, lexical length, historical expanding CTR).

*   **2:00 - 3:00 | Visual Evidence & Results:**
    *   *Baseline vs. Proposed:* Naive Historical CTR Baseline (AUC-ROC: 0.721) vs. LightGBM (AUC-ROC: 0.824).
    *   *Key Chart:* Out-of-time ROC-AUC curve demonstrating a 14.2% relative performance gain.

*   **3:00 - 4:00 | Honest Limitations & Framing:**
    *   *Cold-Start Impact:* Prediction accuracy degrades by 18% on entirely novel terms with zero historical interaction logs.
    *   *Temporal Decay:* Model drift occurs over a 7-day window if features are not refreshed daily.

*   **4:00 - 5:00 | Ranked Action Playbook:**
    *   *Primary Recommendation:* Integrate session-position signals directly into the search candidate reranking layer for immediate cold-start lift; schedule daily rolling re-fits for gradient boosted models.
