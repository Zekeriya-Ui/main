## Mobile UI Issues and Fixes

Based on your description, here are the mobile UI issues identified and the proposed CSS fixes to address them:

### 1. Navigation Links Cluttering on Small Screens

**Problem:** The navigation links become cramped on mobile due to `.site-nav .brand { width: 100%; }` pushing the brand name to its own line, leaving the remaining links (Home, About, Work, Contact) inline with minimal spacing.

**Proposed Fix:** Give `.site-nav` a dedicated full-width sub-grid or flex-row for menu items with equal horizontal spacing and `min-height: 44px` touch targets.

**Conceptual CSS:**
```css
@media (max-width: 768px) { /* Adjust breakpoint as needed */
  .site-nav {
    display: flex;
    flex-direction: column; /* Stack items vertically */
    align-items: center;
  }

  .site-nav .brand {
    width: 100%; /* Ensure brand takes full width */
    text-align: center;
  }

  .site-nav ul {
    display: flex;
    flex-direction: column; /* Stack nav items vertically */
    width: 100%;
    padding: 0;
    list-style: none;
  }

  .site-nav li {
    width: 100%;
    text-align: center;
  }

  .site-nav a {
    display: block;
    padding: 10px 0;
    min-height: 44px; /* Accessible touch target */
    box-sizing: border-box;
  }
}
```

### 2. Mini-Card Grid Over-Compression

**Problem:** On narrower phones, `minmax(220px, 1fr)` causes the 3 stats cards to squeeze unnaturally or wrap in awkward patterns (e.g., 2 then 1).

**Proposed Fix:** Add a mobile breakpoint `@media (max-width: 480px)` that forces `.card-grid` into `grid-template-columns: 1fr` so stats stack cleanly in a vertical list.

**Conceptual CSS:**
```css
@media (max-width: 480px) {
  .card-grid {
    grid-template-columns: 1fr; /* Stack cards vertically */
    gap: 20px; /* Add some spacing between stacked cards */
  }
}
```

### 3. Call-To-Action (CTA) Buttons Squishing

**Problem:** `.cta-row` relies on inline flex wrapping without explicitly setting full-width behavior for buttons on small screens, leading to squished buttons.

**Proposed Fix:** Make `.cta-button` expand to `width: 100%` on screens below `480px` so they act as bold, easy-to-tap primary action bars on phones.

**Conceptual CSS:**
```css
@media (max-width: 480px) {
  .cta-row {
    flex-direction: column; /* Stack CTA buttons vertically */
    gap: 15px; /* Add space between stacked buttons */
  }

  .cta-button {
    width: 100%; /* Make buttons full width */
    box-sizing: border-box; /* Include padding and border in the width */
  }
}
```

### 4. Padding & Font Scaling Micro-Adjustments

**Problem:** Fixed `28px` padding on sections and `3rem` headers consume too much vertical real estate on small smartphone screens.

**Proposed Fix:** Adjust padding dynamically inside the media queries for screens under `640px` and `480px`.

**Conceptual CSS:**
```css
@media (max-width: 640px) {
  section {
    padding: 20px; /* Reduce general section padding */
  }

  h1 {
    font-size: 2.5rem; /* Slightly reduce header size */
  }
}

@media (max-width: 480px) {
  section {
    padding: 15px; /* Further reduce padding for very small screens */
  }

  h1 {
    font-size: 2rem; /* Further reduce header size */
  }

  /* You might also want to adjust other heading levels or paragraph font sizes */
  h2 {
    font-size: 1.5rem;
  }
  p {
    font-size: 0.9rem;
  }
}
```
