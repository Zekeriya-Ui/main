## Resources
Chosen: Road 2 — plain HTML/CSS, built with AI, hosted free on GitHub Pages (same repo as the case study).

I looked at three options: a no-code builder, plain code on a free host, and a full framework with serverless functions. I passed on no-code because it wastes git skill I already have, and it doesn't handle a repo-linked case study or long-form methodology write-up well — it's built for visual portfolios, and mine isn't mostly visual. I passed on the framework-plus-backend route too, even though it's the only one that could show the Signal digest agent live on the site. That's a real capability I'm giving up for now, but the honest trade-off is: it needs a backend just to keep my API key from being exposed in the browser, plus ongoing maintenance — secrets, dependency upgrades, rate-limiting so a stranger can't run up my API bill through my own site. That's more than I need to solve in week 4 of a 10-week build, and the curriculum's own pacing (a dedicated "wire one real thing" week later) backs that up.

Plain HTML/CSS on GitHub Pages is the one option that's actually free, matches what I can build and explain myself, and shows the work the way it needs to be shown: long-form reading for the case study, a repo link that's already sitting there ready to push, and images from the validation charts. The agent doesn't disappear from the story — it just links out for now instead of running live in the page, which is a fine "not yet."

Can I maintain this? Yes — that's the actual reason I picked it. No build tooling to go stale, no dependencies to update, no secrets to manage. If something breaks, it's a plain HTML/CSS file I can read and fix myself, not a framework config I'd have to relearn.
