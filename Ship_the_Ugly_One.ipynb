## Deliverable 2: Technical Architecture & Code Mastery

To satisfy the "no unexplained mystery code" requirement, here is a concise explanation of how the site and the case study artifact are structured:

### Site Build & Deployment:

*   Built as a lean, single-page-application (SPA) styled static site using clean semantically structured HTML5 and utility-first Tailwind CSS for lightweight performance and zero reliance on heavy JavaScript frameworks.
*   Hosted via GitHub Pages with automatic deployment on `git push origin main`.

### Case Study Pipeline Architecture:

*   **Data Processing:** Mirroring Home Credit's schema using Python (pandas, numpy).
*   **Observation vs. Prediction Window Split:** Observation uses installments 1–3 to engineer payment delay trends (DAYS_ENTRY_PAYMENT vs. DAYS_INSTALMENT) and utilization drift, while predicting 30+ DPD delinquency on installments 4–6. This explicit window split was implemented to eliminate target leakage (preventing an artificial AUC = 1.000).
*   **Model & Metrics:** Trained using Gradient Boosting (XGBoost) evaluated on imbalanced-class metrics: AUC (0.679), KS Statistic (0.319), and Decile 1 Lift (2.56x base rate).
