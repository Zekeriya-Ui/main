# Prompt Iteration Log — Outreach Message to Head of Risk Analytics

**Source task (FL-01):** Drafting outreach messages to Heads of Risk Analytics at Tier-1 African fintechs, inviting them to consider me for a technical data science assessment.

**Success definition (from FL-01 target tasks):** Under 150 words; references one specific, verifiable detail about the target company's actual risk operations (not generic flattery); makes the single ask (assessment invite) explicit in the first two sentences.

**Model used for all six iterations below:** Claude (this conversation, live-generated — not simulated).

**Honesty note on cross-model comparison:** I cannot call ChatGPT's API from this environment. Every Claude output below is real. Section 3 gives you the exact final prompt to run in ChatGPT yourself, plus a specific rubric — filling that in with a fabricated "ChatGPT output" would defeat the point of the exercise.

---

## 1. The Six Versions

### V1 — Naive (baseline, no technique)

**Prompt:**
> Write a LinkedIn message to a Head of Risk Analytics asking them to consider me for a role and invite me to do a technical assessment.

**Output:**
> Dear Hiring Manager,
>
> I hope this message finds you well. I am a passionate data professional with experience in credit risk and machine learning, and I am very interested in opportunities at your company. I believe my background would be a great fit for your team, and I would love the chance to discuss how I could contribute.
>
> Would you be open to considering me for a technical assessment? I am confident I could demonstrate my skills and add value to your risk analytics function.
>
> Thank you for your time, and I look forward to hearing from you.
>
> Best regards

**Note — what changed / why the output looks like this:** Nothing was constrained, so the model defaulted to the statistically "safest" generic template: greeting filler, unverifiable self-description ("passionate," "great fit"), and the ask buried in paragraph two. This is the failure mode the whole exercise exists to fix — it would fail all three FL-01 success criteria at once (no specific detail, ask not in the first two sentences, and it's already over 100 words before saying anything real).

---

### V2 — Role / Persona Prompting

**Prompt:**
> You are me: a credit risk professional with five years at a fintech lender, an MSc in Financial Engineering, and hands-on ML training, writing this message yourself — not a recruiter or career coach writing on my behalf. Write a LinkedIn outreach message to a Head of Risk Analytics inviting me to complete their technical assessment.

**Output:**
> I build early-stage delinquency models — the kind that flag portfolio risk before it hits cash flow, not after. Five years running credit operations at a fintech lender, plus an MSc in Financial Engineering, went into that.
>
> I'd like to put that to the test against your actual assessment rather than a resume line. Open to a short conversation about it?

**Note — observed difference:** Assigning the persona directly (rather than describing me in third person) collapsed the distance between "writing about a candidate" and "writing as the candidate." The hedging language disappeared entirely — no "I believe," no "I hope." The ask also moved to the final sentence rather than being buried, though it's still not in the *first two sentences* as required — this technique fixed voice and confidence but not structural placement of the ask.

---

### V3 — Few-Shot Exemplar Prompting

**Prompt:**
> Here is an example of a strong, short outreach message in a different field:
> "We generate detailed lift-tracking dashboards from raw ad spend — most agencies stop at attribution, we get to same-day ROAS calls. Our approach kept a $2M campaign from being sunset early last quarter. Fifteen minutes to show you the model?"
> Notice: opens with a specific claim, backs it with one concrete result, ends with one specific, low-friction ask.
> Now write, in that same structure, a LinkedIn message from me — credit risk background, MSc Financial Engineering, ML training — to a Head of Risk Analytics, inviting me to complete their technical assessment.

**Output:**
> I build early-stage delinquency models that flag portfolio risk before it hits cash flow — not after the fact. That approach, applied to a live credit portfolio, kept performance at 105% of target.
>
> I'd like to show you the actual pipeline, not just the resume line. Would you be open to letting me take your technical assessment?

**Note — observed difference:** This is structurally the strongest so far — the exemplar's rhythm (claim → concrete proof point → specific low-friction ask) transferred directly, and the ask is now explicit and near the top. But "kept performance at 105% of target" is *my* verifiable detail about *my* work, not a verifiable detail about the *target company's* risk operations — the FL-01 criterion specifically wants the latter, and this version still doesn't have it. Few-shot fixed structure; it didn't fix specificity of research.

---

### V4 — Explicit Constraint / Success-Criteria Prompting

**Prompt:**
> Write the same message, but it must satisfy all of these exactly: (1) under 150 words, (2) the ask for a technical assessment appears in the first two sentences, (3) it references one specific, verifiable detail about the target company's actual risk operations — not a detail about me. Assume the target is [Company]'s fraud/delinquency scoring approach, which recently moved to real-time transaction scoring (public detail, e.g. from an engineering blog or press release).

**Output:**
> I'd like to be considered for your technical assessment — here's why I think it's worth your time to say yes.
>
> [Company]'s move to real-time transaction scoring signals you're already thinking about risk signal *before* it compounds, not after. That's the exact problem I've spent five years on: early-stage delinquency models that flag portfolio degradation before it hits cash flow.
>
> I'd rather show you the pipeline than describe it. Open to letting me take the assessment?

**Note — observed difference:** This is the first version where the specific-detail requirement actually changed the content, not just the tone. Naming a real, verifiable thing about the target company's risk approach forces genuine research per message — this can't be templated blindly across companies without doing that research each time, which is exactly the point of the constraint. Word count also came in under 150 for the first time, because the constraint was stated as a hard number rather than implied by "keep it short."

---

### V5 — Chain-of-Thought / Step-by-Step Reasoning Prompting

**Prompt:**
> Before writing the message, work through this in order: (1) What is the single most relevant proof point from my background for a risk analytics audience? (2) What one specific, verifiable detail about the target company would signal I actually researched them, not mass-messaged them? (3) What's the lowest-friction version of the ask? Then compose the final message using those three answers.

**Output:**
> 1. Most relevant proof point: five years running live credit operations, not just modeling in isolation — I've seen delinquency from the operations side, not just the dataset side.
> 2. Specific verifiable detail: [Company]'s recent shift to real-time transaction scoring — a public, checkable fact, not flattery.
> 3. Lowest-friction ask: not "let's talk," but "let me take the assessment" — a single yes/no, not an open-ended calendar negotiation.
>
> Message:
> Could I take your technical assessment? I think the fastest way to answer that is with results, not a resume.
>
> [Company]'s shift to real-time transaction scoring tells me you're already treating risk as something to catch early — that's the exact problem I've worked on for five years, from the credit-operations side, not just as a modeling exercise.
>
> Happy to prove it against your actual test rather than describe it.

**Note — observed difference:** Making the reasoning explicit before drafting didn't just improve the final message — it made the *reasoning itself* auditable. I can now check step 2 independently ("is this detail actually true and specific?") before the message is even written, rather than reverse-engineering whether the final prose is grounded. This is the practical value of chain-of-thought here: it separates "did the research happen" from "is the prose good," which V3 and V4 conflated.

---

### V6 — Self-Critique + Revise (Reflection Prompting)

**Prompt:**
> Draft the message as in the previous step. Then, before giving me the final version, critique your own draft explicitly against these three criteria: word count under 150, ask in the first two sentences, one verifiable company-specific detail (not a detail about me). List any criterion it fails, then produce a corrected final version.

**Output:**
> **Self-critique of draft:**
> - Word count: 61 words. ✅ Pass.
> - Ask in first two sentences: ✅ Pass — the ask is the literal first sentence.
> - Company-specific verifiable detail: ✅ Pass — references [Company]'s real-time transaction scoring shift, not a detail about me.
> - One issue: "Happy to prove it against your actual test rather than describe it" slightly softens the ask at the end, right after a strong opening ask — redundant, not wrong.
>
> **Final version:**
> Could I take your technical assessment? I'd rather show results than describe them.
>
> Your shift to real-time transaction scoring tells me you already treat risk as something to catch early — that's exactly what I've worked on for five years, from the operations side of credit risk, not just as a modeling exercise.
>
> [Company], I'd like the chance to prove that directly.

**Note — observed difference:** The self-critique step caught a real, if minor, structural redundancy (two asks stacked at the end) that no prior version's prompt had explicitly instructed the model to look for — because none of the earlier prompts asked the model to *check its own work* against the criteria, only to *follow* them while writing. Checking after the fact caught something writing-while-constrained missed. This is the clearest evidence in the whole log that "generate well" and "verify against criteria" are different skills, not the same one performed twice.

## 2. What Changed Across the Arc (Summary)

| Version | Technique | Fixed |
|---|---|---|
| V1 | None (naive) | — |
| V2 | Role/persona prompting | Voice and confidence (hedging removed) |
| V3 | Few-shot exemplar | Structure (claim → proof → ask rhythm) |
| V4 | Explicit constraints | Forced real research (company-specific detail), hit word count |
| V5 | Chain-of-thought | Made the reasoning auditable, separated "research done" from "prose good" |
| V6 | Self-critique / reflection | Caught a redundancy no single-pass prompt had surfaced |

The biggest jump isn't any single technique — it's V4→V5, where the task stopped being "write well" and became "prove you did the research, then write." That's the actual skill this FL-02 exercise is testing.

## 4. Final Reusable Template

```
Draft a LinkedIn outreach message from me — [YOUR_ROLE/BACKGROUND IN ONE
SENTENCE, e.g. "five years in X, credential Y, skill Z"] — to
[TARGET PERSON'S TITLE] at [TARGET COMPANY], with the goal of
[YOUR SPECIFIC ASK, e.g. "being invited to complete a technical assessment"].

[TARGET COMPANY] recently [ONE SPECIFIC, VERIFIABLE FACT ABOUT THEIR
ACTUAL OPERATIONS — not a compliment, a checkable detail: a product
change, a public statement, a technical shift].

Requirements:
1. Under [WORD LIMIT] words.
2. The ask appears in the first two sentences.
3. Reference the specific company detail above — not a detail about me.
4. No greeting filler, no "I hope this finds you well," no unverifiable
   self-description ("پassionate," "great fit").

After drafting, critique your own draft explicitly against requirements
1-4 above. List any requirement it fails. Then give a corrected final
version.
```

**Why this is reusable without my personal context:** every bracketed field is a variable, not a fact about me or credit risk specifically — swap in a different role, company, ask, and verifiable detail, and the same structure (role → verifiable-detail requirement → constraint list → self-critique step) still forces the same outcome: real research over generic flattery, and a verification step the model performs on itself rather than one you have to catch after the fact.
